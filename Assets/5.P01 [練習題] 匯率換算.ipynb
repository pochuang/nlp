# 匯率換算
請實作出一個可以進行匯率換算的Tools

```python
# 可進行匯率換算的函數
import requests
from bs4 import BeautifulSoup

def get_rates_xrates(base_currency='TWD'):
    url = f"https://www.x-rates.com/table/?from={base_currency}&amount=1"
    resp = requests.get(url)
    soup = BeautifulSoup(resp.text, 'html.parser')
    tables = soup.find_all("table")
    rates = {}
    for table in tables:
        for tr in table.find_all("tr"):
            tds = tr.find_all("td")
            if tds:
                cur = tds[0].text.strip()
                rate = float(tds[1].text.strip())
                rates[cur] = rate
    return rates

rates = get_rates_xrates('TWD')
print("TWD→USD:", rates['US Dollar'])
print("TWD→JPY:", rates['Japanese Yen'])
```
