# 英文教學對話機器人

## 練習：英文教學對話機器人

### 背景

請利用大型語言模型（LLM）建立一位 **英文教學老師**。學生可以透過多輪對話反覆練習英文翻譯，老師會根據學生的回答給予回饋，並持續出下一題。

---

## 範例對話

**AI**

> 第一題：
>
> 請將下面句子翻譯成英文：
>
> **「我昨天和朋友一起去看電影。」**

---

**學生**

> I go to movie with my friend yesterday.

---

**AI**

> 批改結果：
>
> **拼字**
>
> * `movie` 前建議加上 `the`。
>
> **文法**
>
> * `go` 應改為過去式 `went`。
>
> **更自然的表達**
>
> * `go to the movies` 比 `go to movie` 更自然。
>
> **參考答案**
>
> > I went to the movies with my friend yesterday.
>
> 做得不錯！你已經正確使用了 `yesterday` 表示過去時間，只要注意動詞時態即可。
>
> ---
>
> **下一題：**
>
> 請將下面句子翻譯成英文：
>
> **「如果明天天氣很好，我想去爬山。」**

---

## 思考題

1. 如何設計 System Prompt，避免 AI 一開始就公布答案？
2. 如何讓 AI 每次都按照固定格式批改，而不是每次格式不同？
3. 如何讓 AI 根據學生的程度調整題目難度？
4. 如果學生回答「不知道」或「放棄」，AI 應如何回應？
5. Chat Message List 在這個應用中扮演什麼角色？如果每次對話都重新開始，會對學習體驗造成哪些影響？
