# RAG 練習題：海風記憶博物館導覽員

你是「海風記憶博物館」的數位導覽員開發者。請使用提供的博物館文件建立一套 RAG 問答系統，協助參觀者查詢參觀資訊、展覽內容、活動預約、無障礙服務與研究資料。

## 任務目標

系統必須能根據文件回答問題，並在答案中清楚說明依據的文件名稱。當問題需要整合多份文件時，須整合所有必要資訊；當文件沒有足夠資料、涉及個資，或問題要求推論未被證實的結論時，系統應誠實說明限制。

## 必備功能

1. 匯入所有博物館文件並建立向量資料庫。
2. 使用者提問後，檢索最相關內容並產生繁體中文回答。
3. 每次回答至少列出一份參考文件；跨文件問題須列出所有主要依據。
4. 不可捏造館員姓名、網址、展品細節、即時海況或未提供的規則。
5. 遇到個資問題，例如受訪者住址、電話或未公開訪談內容，應婉拒回答並說明原因。
6. 遇到資料不足的問題，應說明「文件未提供足夠資訊」，並建議使用者詢問服務台或查閱官方即時資訊。

## 測試題目

請使用題庫中的 50 題測試系統，至少完成以下 10 題展示：

1. 星期一有開館嗎？國定假日剛好在星期一又怎麼辦？
2. 臨海市居民何時免費入館？可以同時免費上工作坊嗎？
3. 為什麼〈潮位記錄簿〉不能拍，旁邊螢幕卻能翻？
4. 週末四點導覽和一般導覽差在哪？結束還能慢慢逛嗎？
5. 使用輪椅參觀方便嗎？可以借輪椅嗎？哪些地方比較不適合嬰兒車？
6. 火警和海嘯警報的集合點一樣嗎？
7. 研究生想看完整口述逐字稿，如何申請、多久審核？
8. 潮間帶少四成，等於全市所有魚類少四成嗎？
9. 館內文字和照片能放進報告或 YouTube 嗎？
10. 觀眾記憶與展覽地圖不同，導覽員應直接糾正嗎？
