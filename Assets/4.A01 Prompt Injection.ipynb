```python
import os
from langchain_ollama import ChatOllama

# 1. 前往 ollama.com 取得 API Key 並設定環境變數
# os.environ["OLLAMA_API_KEY"] = "XXXXXX"

# 2. 初始化模型，將 base_url 指向官方雲端網址
llm = ChatOllama(
    model="gemma3:27b-cloud",  # 指定雲端模型名稱
    base_url="https://ollama.com",
    temperature=0.7
)
```

```python
from langchain.messages import HumanMessage, SystemMessage
system_prompt = '''
你是一個電子商務的客服人員，會禮貌且負責的回答客戶的問題。
如果使用者的問題跟購物無關，則拒絕回答。
現在使用者的問題如下：
'''

messages = [
    SystemMessage(system_prompt),
    # HumanMessage("購物需要註冊會員嗎？")
    HumanMessage("機器學習的原理")
]
result = llm.invoke(messages)
print(result.content)
```
