# 旅遊規劃助手

## 練習：旅遊規劃助手

### 背景

請利用大型語言模型（LLM）設計一位 **旅遊規劃助手**。AI 應透過多輪對話逐步了解使用者的需求，並根據對話內容持續調整旅遊建議，而不是每次都重新規劃。

---

## 範例對話

**使用者**

> 我想去日本玩。

**AI**

> 很棒！請問預計旅行幾天？以及大概的預算是多少呢？

---

**使用者**

> 五天，預算四萬元。

**AI**

> 好的，目前我知道：
>
> * 地點：日本
> * 天數：5 天
> * 預算：4 萬元
>
> 請問是自由行還是跟團？另外有沒有特別想去的城市？

---

**使用者**

> 東京自由行，不想去迪士尼。

**AI**

> 了解，我會避開迪士尼行程。
>
> 還想了解一下，你比較喜歡哪些類型的活動？
> 例如：美食、購物、溫泉、自然景點、動漫、歷史文化等。

---

**使用者**

> 喜歡美食和溫泉。

**AI**

> 根據目前的需求，我建議：
>
> * 地點：東京近郊（箱根、川越）
> * 天數：5 天
> * 預算：約 4 萬元
> * 行程特色：美食、溫泉、自由行、不安排迪士尼
>
> （接著提供每日行程建議……）

---

## 思考題

1. 為什麼 AI 不應一次詢問所有旅遊資訊？
2. 如何設計 System Prompt，讓 AI 記住使用者之前提供的需求？
3. 如果使用者中途改變想法（例如改去大阪），Prompt 應如何引導 AI 更新規劃？
4. Chat Message List 在這個應用中扮演什麼角色？如果每次都重新建立對話，會產生哪些問題？
5. 如果希望 AI 在對話最後，整理目前已知的旅遊需求與建議行程，應如何修改 Prompt？
