## 練習一：基本摘要

請撰寫 Prompt，使 AI 能完成以下需求：

* 將會議內容摘要成 **200 字以內**
* 保留重要決策
* 忽略閒聊、抱怨及重複內容

---



## 練習二：重點整理

修改 Prompt，使 AI 依照以下格式輸出：

```text
## 會議摘要

（100~200 字）

## 討論重點

- ...
- ...
- ...

## 最終決議

- ...
- ...
```

---



## 練習三：Action Items

修改 Prompt，整理所有待辦事項。

輸出格式：

| 負責人   | 待辦事項         | 預計完成時間 |
| ----- | ------------ | ------ |
| Bob   | 完成通知中心 API   | 下週三    |
| Frank | 確認 Server 建置 | 今天下午   |
| Alice | 聯絡簡訊廠商       | 今天     |

---



## 練習四：風險分析

請讓 AI 額外整理：

* 專案目前最大的風險有哪些？
* 每項風險造成的影響
* 建議改善方式

---



## 練習五：主管版摘要

假設閱讀者是公司主管。

請讓 AI：

* 只保留主管需要知道的資訊
* 不超過 150 字
* 不列出細節討論
* 強調進度、風險、決策
