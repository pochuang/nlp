# 猜人物遊戲

## 練習：猜人物遊戲（Who Am I?）

### 背景

請利用大型語言模型（LLM）設計一個 **猜人物遊戲**。AI 會先在心中選定一位知名人物，玩家透過多輪對話提出問題，AI 只能回答有限的資訊，直到玩家猜出人物為止。

---

## 範例對話

**AI**

> 我已經選好一位知名人物了！
> 你可以開始提問，每次問一個問題，我會根據人物資訊回答。

---

**玩家**

> 他是男生嗎？

**AI**

> 是。

---

**玩家**

> 他是運動員嗎？

**AI**

> 不是。

---

**玩家**

> 他是台灣人嗎？

**AI**

> 是。

---

**玩家**

> 他還在世嗎？

**AI**

> 是。

---

**玩家**

> 他是周杰倫嗎？

**AI**

> 🎉 恭喜你答對了！答案就是 **周杰倫**。
> 想再玩一局嗎？

---

## 思考題

1. 如何設計 System Prompt，避免 AI 一開始就洩漏答案？
2. 如果玩家問「請直接告訴我答案」，如何讓 AI 仍維持遊戲規則？
3. 如何讓 AI 對於模糊或無法回答的問題（例如「他很有名嗎？」）做出適當回應？
4. Chat Message List 在這個遊戲中扮演什麼角色？如果沒有保存對話紀錄，會對遊戲造成哪些影響？
5. 如果希望 AI 能統計玩家已提問的次數，並在每回合提醒「目前已提問 X 次」，應如何利用 Chat Message List 與 Prompt 來實現？
