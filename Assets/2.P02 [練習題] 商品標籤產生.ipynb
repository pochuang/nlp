# 商品標籤產生

## 練習：商品標籤自動產生

### 背景

某電商平台希望利用大型語言模型（LLM）協助商品上架。請根據商品描述，從提供的**商品標籤列表**中挑選最適合的標籤，以提升商品搜尋與分類品質。

---

### 任務

請設計 Prompt，讓 LLM 根據商品描述，完成商品標籤推薦。

### 規則

1. **只能從提供的商品標籤列表中選擇標籤。**
2. 每個商品請選出 **3～5 個**最適合的標籤。
3. 不可新增標籤或修改標籤名稱。
4. 不可重複選擇相同標籤。
5. 請以 JSON Array 格式輸出。

---

### 範例

**商品描述**

> 雙層真空不鏽鋼保溫瓶，容量 750ml，可保溫 12 小時、保冰 24 小時，適合露營、健行及日常通勤。

**輸出**

```json
[
    "配件",
    "保溫",
    "保冰",
    "不鏽鋼",
    "露營"
]
```

### 思考題

1. 如果沒有告訴 LLM「只能從標籤列表中選擇」，可能會產生什麼問題？
2. 如何修改 Prompt，避免模型輸出不存在的標籤？
3. 如果希望每個商品**一定包含一個「商品類別」標籤**，Prompt 應如何設計？
4. 如果希望所有輸出都符合 JSON 格式，可以加入哪些 Prompt 限制？
5. 若同一商品有多種用途（例如露營、旅行、通勤），應如何引導 LLM 選出最具代表性的標籤？
